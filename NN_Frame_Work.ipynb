```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)
```

```python
x1 = 3
x2 = 2
y_actual = 17
```

```python
w1 = random.uniform(1, 10)
w2 = random.uniform(1, 10)

print("Initial Weights")
print("w1 =", round(w1, 4))
print("w2 =", round(w2, 4))
```

```python
# Learning Rate
learning_rate = 0.01

# Lists to store values
errors = []
predictions = []
```

```python
epochs = 20

for epoch in range(epochs):

    # Feed Forward
    y_pred = (x1 * w1) + (x2 * w2)

    # Error Calculation
    error = (y_actual - y_pred) ** 2

    # Store results
    errors.append(error)
    predictions.append(y_pred)

    # Back Propagation
    grad_w1 = -2 * (y_actual - y_pred) * x1
    grad_w2 = -2 * (y_actual - y_pred) * x2
```

```python
w1 = w1 - learning_rate * grad_w1
w2 = w2 - learning_rate * grad_w2

print(f"Epoch {epoch+1:2d} | Predicted = {y_pred:.4f} | Error = {error:.4f}")
```

```python
print("\nFinal Weights")
print("w1 =", round(w1, 4))
print("w2 =", round(w2, 4))

print("\nFinal Predicted Value =", round(predictions[-1], 4))
print("Actual Value =", y_actual)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1, epochs+1), errors, marker='o')
plt.title("Error vs Epochs")
plt.xlabel("Epoch")
plt.ylabel("Squared Error")
plt.grid(True)
plt.show()
```

```python
# Predicted Output Progression
plt.figure(figsize=(8,5))
plt.plot(range(1, epochs+1), predictions, marker='o')
plt.axhline(y=y_actual, linestyle='--', label='Actual Value = 17')
plt.title("Predicted Output vs Epochs")
plt.xlabel("Epoch")
plt.ylabel("Predicted Output")
plt.legend()
plt.grid(True)
plt.show()
```
